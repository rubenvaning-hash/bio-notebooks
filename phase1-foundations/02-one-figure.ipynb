# 02: One figure

**Claim:** no single measurement keeps the three penguin species apart, but two measurements together can.

```python
import seaborn as sns
import matplotlib.pyplot as plt
df = sns.load_dataset('penguins')
```

Prediction: on its own, bodymass cannot split all species

```python
boxplot = sns.boxplot(x='species', y='body_mass_g', data=df)
```

Prediction: bill_length splits Adelie from Chinstrap. 

```python
sns.scatterplot(x='bill_length_mm', y='body_mass_g', hue='species', data=df)
```

Prediction: bill depth does separate Gentoo and Chinstrap

```python
sns.scatterplot(x='bill_length_mm', y='bill_depth_mm', hue='species', data=df)
```

**Observation (not a blind prediction, seen in the scatterplots above):** bill length alone cannot separate Chinstrap from Gentoo.

**Why panel A is bill length and not body mass:** panel B then adds exactly one measurement to panel A, and bill length x bill depth leaves fewer overlapping points than bill length x body mass (compare the two scatterplots above).

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
axes[0].set_title('A: Bill length alone cannot separate Chinstrap from Gentoo')
sns.stripplot(x='species', y='bill_length_mm', hue='species', data=df, ax=axes[0], legend=False)
axes[1].set_title('B: Adding bill depth mostly separates all three species')
sns.scatterplot(x='bill_length_mm', y='bill_depth_mm', hue='species', data=df, ax=axes[1])
```

**Finding:** bill length alone cannot separate Chinstrap from Gentoo (both span roughly 41 to 58 mm), but adding bill depth mostly does, because Gentoo have shallower bills (13 to 17 mm vs 16 to 21 mm for Chinstrap); only a handful of Adelie and Chinstrap penguins still overlap, around 42 to 45 mm bill length.
